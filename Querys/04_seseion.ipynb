Ejercicio 17: ¿los hijos aseguran el viaje?

```python
%sql
SELECT CASE WHEN children > 0 OR babies > 0 THEN 'Con Hijos' ELSE 'Sin Hijos' END AS segmento,
       COUNT(*) AS total_reservas,
       ROUND(AVG(is_canceled), 4) AS tasa_cancelacion
FROM cursodataacademy.proyecto_hotelero.hotel_bookings
GROUP BY 1
ORDER BY tasa_cancelacion DESC;
```

Ejercicio 18: solicitudes especiales

```python
%sql
SELECT total_of_special_requests,
       COUNT(*) AS total_reservas,
       ROUND(AVG(is_canceled), 4) AS tasa_cancelacion
FROM cursodataacademy.proyecto_hotelero.hotel_bookings
GROUP BY total_of_special_requests
ORDER BY total_of_special_requests ASC;
```

Ejercicio 19: cambios en la reserva

```python
%sql
SELECT CASE WHEN booking_changes = 0 THEN 'Sin Cambios'
            WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
            ELSE 'Más de 3 cambios' END AS grupo_cambios,
       COUNT(*) AS total_reservas,
       ROUND(AVG(is_canceled), 4) AS tasa_cancelacion
FROM cursodataacademy.proyecto_hotelero.hotel_bookings
GROUP BY 1
ORDER BY tasa_cancelacion ASC;
```

Ejercicio 20: factor precio

```python
%sql
SELECT hotel,
       COUNT(*) AS total_cancelaciones,
       ROUND(AVG(adr), 2) AS adr_promedio_canceladas
FROM cursodataacademy.proyecto_hotelero.hotel_bookings
WHERE is_canceled = 1
GROUP BY hotel
ORDER BY adr_promedio_canceladas DESC;
```

```python
import os
import pandas as pd

# 1. RUTA: Apuntando a tu carpeta en el Workspace
folder_path = (
    "/Workspace/Users/u23251675@utp.edu.pe/Proyecto-SQL-Hotelbookings-JAPA"
)

# 2. ESQUEMA: Destino en Databricks (sin barras diagonales al final)
catalog_schema = "cursodataacademy.proyecto_hotelero"

# Buscamos los archivos CSV en la carpeta
csv_files = [f for f in os.listdir(folder_path) if f.lower().endswith(".csv")]

print(f"Se encontraron {len(csv_files)} archivos CSV:\n")

for csv_file in sorted(csv_files):
  file_path = os.path.join(folder_path, csv_file)
  table_name = csv_file.replace(".csv", "").lower()
  full_table_name = f"{catalog_schema}.{table_name}"

  # 3. MANEJO DE SEPARADORES: Ajusta el delimitador según tus archivos (coma por defecto)
  separador = ";" if "ticket" in csv_file.lower() else ","

  # Leer CSV con el separador dinámico y codificación UTF-8
  df_pandas = pd.read_csv(
      file_path, encoding="utf-8-sig", sep=separador, skipinitialspace=True
  )

  # Reemplaza los espacios y caracteres problemáticos por guiones bajos en los nombres de columnas
  df_pandas.columns = df_pandas.columns.str.replace(
      r"[ ,.;(){}\n\t-]", "_", regex=True
  )

  # Crear DataFrame de Spark y guardar como tabla Delta
  df_spark = spark.createDataFrame(df_pandas)
  df_spark.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable(
      full_table_name
  )

  row_count = df_spark.count()
  col_count = len(df_spark.columns)
  print(
      f"{csv_file} -> {full_table_name} ({row_count} filas, {col_count}"
      f" columnas: {df_spark.columns})"
  )

print(f"\n Carga completa: {len(csv_files)} tablas creadas en {catalog_schema}")
```

Ejercicio 11 (Costo del marketing inútil)

```python
%sql
SELECT
  COUNT(*)        AS reservas_canceladas_ta_to,
  COUNT(*) * 1.5  AS perdida_total_usd
FROM cursodataacademy.proyecto_hotelero.tb_reservas_prueba
WHERE distribution_channel = 'TA/TO'
  AND is_canceled = 1;
```

Ejercicio 12 (Costo de cancelar tarde)

```python
%sql
SELECT
  COUNT(*)        AS cancelaciones_tardias,
  COUNT(*) * 120  AS perdida_total_usd
FROM cursodataacademy.proyecto_hotelero.tb_reservas_prueba
WHERE is_canceled = 1
  AND DATEDIFF(
    to_date(CONCAT_WS(' ', arrival_date_day_of_month, arrival_date_month, arrival_date_year), 'd MMMM y'),
    to_date(reservation_status_date, 'd/M/yy')
  ) < 3;
```

Ejercicio 14 (INNER JOIN con agrupación)

```python
%sql
SELECT p.nombre_pais, COUNT(*) AS reservas_canceladas
FROM cursodataacademy.proyecto_hotelero.hotel_bookings r
INNER JOIN cursodataacademy.proyecto_hotelero.tb_paises p
  ON r.country = p.cod_pais
WHERE r.is_canceled = 1
GROUP BY p.nombre_pais
ORDER BY reservas_canceladas DESC
LIMIT 5;
```

Ejercicio 15 (FULL OUTER JOIN)

```python
%sql
CREATE OR REPLACE TABLE cursodataacademy.proyecto_hotelero.tb_vendedores AS
SELECT * FROM VALUES
  (1,'Ana Quispe'),(2,'Luis Rojas'),(3,'María Flores'),(4,'Carlos Mamani'),(5,'Rosa Huamán')
AS t(id_vendedor, nombre_vendedor);
```

```python
%sql
CREATE OR REPLACE TABLE cursodataacademy.proyecto_hotelero.tb_reservas_vendedor AS
SELECT
  distribution_channel,
  CASE WHEN distribution_channel = 'TA/TO'
       THEN CAST(pmod(ROW_NUMBER() OVER (ORDER BY arrival_date_year, arrival_date_week_number, lead_time), 4) + 1 AS INT)
  END AS id_vendedor
FROM cursodataacademy.proyecto_hotelero.hotel_bookings;
```

```python
%sql
SELECT r.distribution_channel, v.nombre_vendedor
FROM cursodataacademy.proyecto_hotelero.tb_reservas_vendedor r
FULL OUTER JOIN cursodataacademy.proyecto_hotelero.tb_vendedores v
  ON r.id_vendedor = v.id_vendedor;
```

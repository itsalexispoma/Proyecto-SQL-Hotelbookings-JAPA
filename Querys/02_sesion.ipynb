Ejercicio 1 (CREATE / INSERT)

```python
%sql
-- Se usa una tabla Delta de prueba (permite UPDATE y DELETE)
CREATE OR REPLACE TABLE tb_reservas_prueba AS
SELECT * FROM hotel_bookings WHERE 1 = 0;   -- copia solo la estructura

INSERT INTO tb_reservas_prueba
SELECT *
FROM hotel_bookings
WHERE hotel = 'City Hotel'
  AND is_canceled = 1;
```

Ejercicio 2 (UPDATE)

```python
%sql
UPDATE tb_reservas_prueba
SET babies = 0;
```

Ejercicio 3 (CAST / CONCAT)

```python
%sql
SELECT
  CONCAT_WS('-',
    LPAD(CAST(arrival_date_day_of_month AS STRING), 2, '0'),
    arrival_date_month,
    CAST(arrival_date_year AS STRING)
  ) AS arrival_date_string
FROM hotel_bookings;
```

Ejercicio 4 (LIKE)


```python
%sql
SELECT *
FROM tb_reservas_prueba
WHERE distribution_channel LIKE 'T%';
```

Ejercicio 5 (COUNT / SUM / AVG)

```python
%sql
SELECT
  COUNT(*)                          AS total_reservas,
  SUM(total_of_special_requests)    AS total_solicitudes_especiales,
  ROUND(AVG(adr), 2)                AS adr_promedio
FROM tb_reservas_prueba;
```

Ejercicio 6 (DELETE)

```python
%sql
DELETE FROM tb_reservas_prueba
WHERE adr = 0.0;
```

Ejercicio 7 (GROUP BY)

```python
%sql
SELECT
  distribution_channel,
  COUNT(*) AS cantidad_reservas
FROM tb_reservas_prueba
GROUP BY distribution_channel
ORDER BY cantidad_reservas DESC;
```

Ejercicio 8 (CASE WHEN)

```python
%sql
SELECT
  adr,
  CASE
    WHEN adr < 50              THEN 'Económico'
    WHEN adr BETWEEN 50 AND 150 THEN 'Estándar'
    ELSE 'Premium'
  END AS categoria_precio
FROM tb_reservas_prueba;
```

Ejercicio 9 (DATE_SUB)

```python
%sql
SELECT
  to_date(
    CONCAT_WS(' ', arrival_date_day_of_month, arrival_date_month, arrival_date_year),
    'd MMMM y'
  ) AS arrival_date,
  lead_time,
  DATE_SUB(
    to_date(CONCAT_WS(' ', arrival_date_day_of_month, arrival_date_month, arrival_date_year), 'd MMMM y'),
    CAST(lead_time AS INT)
  ) AS booking_date
FROM tb_reservas_prueba;
```

Ejercicio 10 (Funciones matemáticas)

```python
%sql
SELECT
  lead_time,
  CEIL(lead_time / 30) AS meses_anticipacion_techo
FROM tb_reservas_prueba;
```
